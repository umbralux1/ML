```python
import pandas as pd
import matplotlib.pyplot as plt
from sklearn.linear_model import LinearRegression
```

```python
data = pd.read_csv('movies_budget_gross.csv')
data.describe()
```

```python
model = LinearRegression()
X = pd.DataFrame(data.budget_musd)
y = pd.DataFrame(data.worldwide_gross_musd)
model.fit(X, y)
```

```python
model.predict(pd.DataFrame({'budget_musd': [45]}))
```

```python
plt.figure(figsize=(10, 6))
plt.scatter(data.budget_musd, data.worldwide_gross_musd, alpha=0.7)
plt.xlabel('Бюджет')
plt.ylabel('Кассовый сбор')
#plt.ylim(10, 2923)
#plt.xlim(0.0009, 379)
plt.plot(X, model.predict(X), color='red', linewidth=1.4)
```

```python
model.score(X, y)
```
