```python
import pandas as pd, numpy as np
from sklearn.linear_model import LinearRegression
import matplotlib.pyplot as plt
```

```python
df = pd.read_csv('movies_budget_gross.csv')
x = df[['budget_musd']]
y = df['worldwide_gross_musd']
m = LinearRegression().fit(x, y)
print(m.coef_[0], m.intercept_, m.score(x, y))
```

```python
m.predict(pd.DataFrame({'budget_musd': [150]}))
```

```python
plt.figure(figsize=(10, 6))
plt.scatter(df['budget_musd'], df['worldwide_gross_musd'], alpha=0.7, label='Фильмы')
xs = np.linspace(0, df['budget_musd'].max(), 100)
plt.plot(xs, m.predict(pd.DataFrame({'budget_musd': xs})), color='red', label='Регрессия')

plt.xlabel('Бюджет, млн $')
plt.ylabel('Мировые сборы, млн $')
plt.title('Зависимость сборов от бюджета')
plt.legend()
plt.grid(alpha=0.3)
plt.show()
```
